```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
import pandas as pd
k=[]
for col in data.columns:
  if (data[col].dtype !='object') & (col!='price'):
    k.append(col)
x=data[k]
vif=pd.DataFrame()
vif['features']=x.columns
vif["VIF"] = [variance_inflation_factor(x.values, i) for i in range(x.shape[1])]
print(vif)
```

```python
data=data.drop(columns=['flight'])
```

```python
x=data.iloc[:,:-1] # tell us all features except last one(input value)
y=data['price']    #(we have to predict i.e  our target variable)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing  import StandardScaler
```

```python
s=StandardScaler()
```

```python
x_train=s.fit_transform(x_train)
x_test=s.transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
l=LinearRegression()
```

```python
l.fit(x_train,y_train)
```

```python
y_pred=l.predict(x_test)
```

```python
difference=pd.DataFrame(np.c_[y_test,y_pred],columns=['Actual_Value','Predicted_Value'])
difference
```

```python
from sklearn.metrics import *
from sklearn import metrics
```

```python
r2_score(y_test,y_pred)
```

```python
sns.distplot(y_test,label='Actual')
sns.distplot(y_pred,label='Predicted')
plt.legend()
```

```python
mean_abs_error=metrics.mean_absolute_error(y_test,y_pred)
mean_abs_error
```

```python
mean_absolute_percentage_error(y_test,y_pred)
```

```python
root_mean_sq_err=root_mean_squared_error(y_test,y_pred)
root_mean_sq_err
```
